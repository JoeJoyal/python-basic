The **area of a rectangle** is the amount of space inside the rectangle.



**formula:** Area = Length * Width

**Calculate the Area of a Rectangle**

```python
length = 10
width = 5

area = length * width

print("length:", length)
print("width:", width)
print("Area:", area)
```

**Using User Input**

```python
length = float(input("Enter the length: "))
width = float(input("Enter the width: "))

area = length * width

print("Area of the rectangle: ", area)
```

**Using Decimal Values**

```python
length = 5.5
width = 3.2

area = length * width

print("Area:", area)
```

```python
length = 12
width = 8

area = length * width

print("Room Area:", area, "square metres")
```

The `*` operator performs multiplication, so multiplying the length by the width gives the area of the rectangle.
