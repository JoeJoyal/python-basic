**Convert a Number String to a Float**

```python
number_string = "25.75"

number = float(number_string)

print(number)
print(type(number))
```

```python
price = "99.99"

price = float(price)

print(price)
```

**Using User Input**

```python
number_string = input("enter a decimal number :")

number = float(number_string)

print("Float value :", number)
print(type(number))
```
