**Using a Temporary Variable**

```python
a = 10
b = 20

print("Before swapping:")
print("a =", a)
print("b =", b)

temp = a
a = b
b = temp

print("After swapping:")
print("a =", a)
print("b =", b)
```

**Using Python Multiple Assignment**

```python
a = 10
b = 20

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

**Using Addition and Subtraction**

```python
a = 10
b = 20

a = a + b
b = a - b
a = a - b

print(a)
print(b)
```

**Using Multiplication and Division**

```python
a = 10
b = 20

a = a * b
b = a / b
a = a / b


print(int(a))
print(int(b))
```

**Using Bitwise XOR**

```python
a = 10
b = 20

a = a ^ b
b = a ^ b
a = a ^ b

print(a)
print(b)
```
