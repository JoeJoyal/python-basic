**Direct Assignment**

```python
age = 25
print(age)
```

**Using User Input**

```python
age = int(input("Enter your age: "))
print(age)
```

**Using `int()`**

```python
age = int(30)
print(age)
```

```python
type(age)
```

**Using a Variable**

```python
my_age = 25
print(my_age)
```

**Using an f-string**

```python
age = 25
print(f"My age is {age}")
```

**Using String Concatenation**

```python
age = 25
print("My age is " + str(age))
```

**Using a Dictionary**

```python
person = {
    "age":25
}
print(person['age'])
```

**Using a Tuple**

```python
data = (25,)
print(data[0])
```

**Using a Function**

```python
def get_age():
    return 25

age = get_age()
```

```python
print(age)
```

```python
type(get_age())
```

**Using a Class**

```python
class Person:
    def __init__(self, age):
        self.age = age

person = Person(25)
```

```python
print(person.age)
```
