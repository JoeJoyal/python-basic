**Using `Str()` Conversion**

```python
name = "John"
age = 25

sentence = "My name is " + name + " and I am " + str(age) +  " years old. "
print(sentence)
```

**Using f-string**

```python
name = "John"
age = 25

sentence = f"My name is {name} and I am {age} years old."
print(sentence)
```

**Using `format()` Method**

```python
name = "John"
age = 25

sentence = "My name is {} and I am {} years old.".format(name, age)
print(sentence)
```

**Using Multiple Data Types**

```python
name = "John"
age = 25
height = 5.8

sentence = f"My name is {name}, I am {age} years old, and my height is {height} feet."
print(sentence)
```

**Using `print()` with Multiple Values**

```python
name = "John"
age = 25

print("My name is", name, "and I am ", age, "Years old.")
```
