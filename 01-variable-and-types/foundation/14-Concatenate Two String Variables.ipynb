**Using the `+` Operator**

```python
first_name = "John"
last_name = "Doe"

full_name = first_name + last_name

print(full_name)
```

**Concatente with a Space**

```python
first_name = "John"
last_name = "Doe"

full_name = first_name +" "+ last_name

print(full_name)
```

**Using an f-string**

```python
first_name = "John"
last_name = "Doe"

full_name = f"{first_name} {last_name}"

print(full_name)
```

**Using the `join()` Method**

```python
first_name = "John"
last_name = "Doe"

full_name = " ".join([first_name, last_name])
print(full_name)
```

**Using User Input**

```python
first_name = input("Enter your first name :")
last_name = input("Enter your last name: ")

full_name = first_name +" "+ last_name

print("Full name:", full_name)
```
