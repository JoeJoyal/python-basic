**Convert a Flat to an Integer**

```python
number = 25.99

integer_number = int(number)

print(integer_number)
print(type(integer_number))
```

```python
price = 99.75

price = int(price)

print(price)
```

**Using User Input**

```python
number = float(input("Enter a decimal number :"))

integer_number = int(number)

print("Integer value:", integer_number)
print(type(integer_number))
```

Important Difference: **int()** vs **round()**

|Float Value|int() Result|	round() Result|
| ----| ----| ----|
|7.9|7|8|
|5.2|	5|	5|
|10.99|	10|	11|
|-3.8|	-3|	-4|

- `int()` removes the decimal part (truncation).
- `round()` rounds to the nearest whole number.
