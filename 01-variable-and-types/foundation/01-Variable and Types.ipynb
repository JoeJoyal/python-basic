```python
name = "".join(["J", "o", "h", "n"])
print(name)
```

**Using an f-string**

```python
first ="John"

name = f"{first}"
print(name)
```

**From a list**

```python
letters = ['J', 'o', 'h', 'n']

name = "".join(letters)
print(name)
type(name)
```
