**Store a Boolean Value**

```python
is_student = True

print(is_student)
print(type(is_student))
```

**Flip a Boolean Using `not`**

```python
is_student = True

is_student = not is_student

print(is_student)
```

**Flip Multiple Times**

```python
status = True

print(status)

status = not status
print(status)

status = not status
print(status)
```

**Real Time example**

```python
is_logged_in = False

print("Before login:", is_logged_in)

is_logged_in = not is_logged_in
print("After login:", is_logged_in)
```

**Flip using a function**

```python
def flip_boolean(value):
    return not value

light_on = True

light_on = flip_boolean(light_on)

print(light_on)
```

This technique is commonly used for switches, login states, visibility toggles, and enable/disable flags in Python programs.
