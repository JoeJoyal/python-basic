**Using `round()`**

```python
number = 3.14159

rounded_number = round(number, 2)

print(rounded_number)
```

```python
price = 99.98765

price = round(price, 2)

print(price)
```

**Using an f-string**

```python
number = 8.56789

print(f"{number:.2f}")
```

**Using `format()`**

```python
number = 5.67891

print("{:.2f}".format(number))
```

**User Input example**

```python
number = float(input("Enter a decimal number:"))

rounded = round(number, 2)

print("Rounded value:", rounded)
```
