**Calculate Total Cost**

```python
price = 50
quantity = 3

total_cost = price * quantity

print("Price:", price)
print("Quantity:", quantity)
print("Total Cost:", total_cost)
```

**Using Float Values (Demical Cost)**

```python
item_price = 25.50
quantity = 4

total_cost = item_price * quantity

print("Item Price:", item_price)
print("Quantity:", quantity)
print("Total Cost:", total_cost)
```

**Shopping Bill example**

```python
laptop_price = 550000
mouse_price = 800
keyboard_price = 1500

total_cost = laptop_price + mouse_price + keyboard_price

print("Lapout: ", laptop_price)
print("Mouse: ", mouse_price)
print("Keyboard :", keyboard_price)
print("Total bill :", total_cost)
```

**With Tax Calculation**

```python
product_price = 1000
tax_rate = 0.18

tax = product_price * tax_rate
total_cost = product_price + tax

print("Product Price:", product_price)
print("Tax:", tax)
print("Total Cost: ", total_cost)
```

**Formula :** total_cost = price * quantity

Where:

`price` → Cost of one item

`quantity` → Number of items purchased

`total_cost` → Final amount to pay

This concept is commonly used in **billing systems, shopping carts, invoices, and financial applications.**
