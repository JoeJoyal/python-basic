**Count Characters in a String**

```python
name = "Python"

count = len(name)

print("String:", name)
print("No.of characters: ", count)
```

**Count Characters Including Spaces**

```python
message = "Hello World"

count = len(message)

print(count)
```

**Count Characters From User Input**

```python
text = input("Enter a string:")

length = len(text)

print("Total characters:", length)
```

**Count Characters Without Spaces**

```python
text = "Hello World"

count = len(text.replace(" ", ""))
print(count)
```

**Count Frequency of Each Character**

```python
text = "Python"

for char in text:
    print(char, "=", text.count(char))
```

**Using Dictionary to Count Characters**

```python
text = "hello"

char_count = {}

for char in text:
    if char in char_count:
        char_count[char] +=1
    else:
        char_count[char] = 1

print(char_count)
```
