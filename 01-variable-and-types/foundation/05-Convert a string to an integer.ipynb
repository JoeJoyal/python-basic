**Convert a String to an Integer**

```python
num_str = "25"

num = int(num_str)
print(num)
print(type(num))
```

**Using User Input**

```python
age = input("Enter your age: ")
age = int(age)

print("Your age is:", age)
print(type(age))
```

**Convert Multiple Strings**

```python
a = "10"
b = "20"

sum = int(a) + int(b)
print(sum)
```

**Invalid Conversion**

If the string is not a valid integer, Python raises a `ValueError`.

```python
num = "Hello"

num = int(num)
```

**Handle Invalid Input with `try-except`**

```python
num_str = input("Enter a number: ")

try:
    num = int(num_str)
    print("Integer:", num)
except ValueError:
    print("Invalid input! Please enter a valid integer.")
```
