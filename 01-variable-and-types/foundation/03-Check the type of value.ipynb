**Check the type of an Integer**

```python
age = 25
print(type(age))
```

**Check the type of String**

```python
name = "John"
print(type(name))
```

**Check the type of a Float**

```python
height = 5.8
print(type(height))
```

**Check the type of a Boolean**

```python
is_student = True
print(type(is_student))
```

**Check the type of Multiple Value**

```python
name = "John"
age = 30
height = 5.8
is_student = True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```
