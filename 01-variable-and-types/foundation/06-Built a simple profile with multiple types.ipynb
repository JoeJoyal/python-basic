**Simple Profile Using Variable**

```python
name = "John"
age = 25
height = 5.8
weight = 70.5
is_student = True
city = "Bangalore"

print("----- My Profile -----")
print("Name", name)
print("Age", age)
print("Height", height)
print("Weight", weight)
print("Student", is_student)
print("City", city)
```

**Profile Using a Dictionary**

```python
profile = {
    "name": "John",
    "age":25,
    "height": 5.8,
    "weight": 70.5,
    "is_student": True,
    "skills": ['Python', 'SQL', 'Machine Learning']
}

print(profile)
```

**Print Each Field**

```python
profile = {
    "name": "John",
    "age":25,
    "height": 5.8,
    "weight": 70.5,
    "is_student": True,
    "skills": ['Python', 'SQL', 'Machine Learning']
}
```

```python
print("Name", profile['name'])
print("Age", profile['age'])
print("Hieght", profile['height'])
print("Weight", profile['weight'])
print("Student", profile['is_student'])
print("Skills", profile['skills'])
```
