**Convert an Integer to a Float**

```python
num = 25

result = float(num)

print(result)
print(type(result))
```

**Using User Input**

```python
num = int(input("Enter an integer: "))

result = float(num)

print(result)
print(type(result))
```

**Convert Multiple Integers**

```python
a = 10
b = 20

x = float(a)
y = float(b)

print(x)
print(y)
print( ( type(x), type(y) ) )
```

**Perform Calculations After Conversion**

```python
num = 15

result = float(num)

print(result + 2.5)
```
