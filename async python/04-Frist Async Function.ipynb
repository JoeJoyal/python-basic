**Normal Python function**

````python
def hello():
    print("hello")

hello()
````

````python
print(hello())
````

**Your first async function**

Add `async` before `def`:

````python
import asyncio
````

````python
async def hello():
    print("hello")
    
await hello()
````

What changed?

**def hello():**

means: Normal/synchronous function

while: **async def hello():**

means: This is an asynchronous function (coroutine function)

Python doesn't immediately execute the funtion body. It creates a **coroutine object.**

```
async def hello()
        ↓
     hello()
        ↓
  Coroutine object
        ↓
      await
        ↓
   Function executes
```

**Add an asynchronous wait**

Now let's make it more interesting:

````python
async def hello():
    print("hello")

    await asyncio.sleep(2)
    
    print("goodbye")

await hello()
````

```
Hello
    ⏳ 2 seconds
Goodbye
```

**Two async functions**

````python
async def task1():
    print("Task 1 started: ")
    await asyncio.sleep(2)
    print("Task 1 done")

async def task2():
    print("Task 2 started: ")
    await asyncio.sleep(2)
    print("Task 2 done")

await asyncio.gather(task1(), task2())
````

### **🧠 The 3 keywords to remember**

`async` - Create an asynchronous function **async def task()**

`await` - Wait for an asynchronous operation **While giving control back to the event loop `await.asyncio.sleep(2)`**

`asyncio.gather()` - Run multiple async task concurrently

**await asyncio.gather(
    task1()
    task2()
)**

````python
import asyncio
import time

async def task1():
    print("Task 1 started")
    await asyncio.sleep(2)
    print("Task 1 done")

async def task2():
    print("Task 2 started")
    await asyncio.sleep(1)
    print("Task 2 done")

start_time = time.time()

await asyncio.gather(task1(), task2())

end_time = time.time()

print(f"Elapsed: {int(end_time - start_time)} second(s)")
````
