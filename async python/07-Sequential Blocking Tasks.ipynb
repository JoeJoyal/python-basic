````python
import time

def task1():
    print("Task 1 started")
    time.sleep(2)
    print("Task 1 done")

def task2():
    print("Task 2 started")
    time.sleep(2)
    print("Task 2 done")

def main():
    task1()
    task2()

main()
````

⏱️ Total time ≈ 4 seconds

**Execution workflow**

```
                 main()
                   │
                   ▼
             ┌───────────┐
             │  Task 1   │
             └─────┬─────┘
                   │
                   ▼
             sleep(2) ⏳
                   │
                   │ 🔴 BLOCKING
                   ▼
             Task 1 done
                   │
                   ▼
             ┌───────────┐
             │  Task 2   │
             └─────┬─────┘
                   │
                   ▼
             sleep(2) ⏳
                   │
                   │ 🔴 BLOCKING
                   ▼
             Task 2 done
```


**Timeline**

```
Time →     0 sec        2 sec        4 sec
           │             │             │
           ▼             ▼             ▼

Task 1    ├──────────────┤
          sleep(2)       done

Task 2                   ├──────────────┤
                         sleep(2)       done

           ← 2 sec →     ← 2 sec →
           
           Total = 4 seconds
```

**Why is this called sequential?**

Look at these two lines:

```
task1()
task2()
```

Python executes them from top to bottom.
```
task1()
  ↓
FINISH
  ↓
task2()
  ↓
FINISH

```

Task 2 cannot start until Task 1 returns.

**Add execution-time measurement**

````python
import time

def task1():
    print("Task 1 started")
    time.sleep(2)
    print("Task 1 done")

def task2():
    print("Task 2 started")
    time.sleep(2)
    print("Task 2 done")

start_time = time.time()

task1()
task2()

end_time = time.time()

elapsed = int(end_time - start_time)

print(f"Elapsed: {elapsed} second(s)")
````

**Compare with asynchronous execution**

🔴 Sequential blocking
```
Task 1 ────────── 2 sec
                  ↓
Task 2            ────────── 2 sec

Total = 4 sec
```

🟢 Concurrent async
```
Task 1 ────────── 2 sec
Task 2 ────────── 2 sec

Total ≈ 2 sec
```

**Async version:**

````python
import asyncio

async def task1():
    print("Task 1 started")
    await asyncio.sleep(2)
    print("Task 1 done")

async def task2():
    print("Task 2 started")
    await asyncio.sleep(2)
    print("Task 2 done")

async def main():
    await asyncio.gather(task1(), task2())

await main()
````
