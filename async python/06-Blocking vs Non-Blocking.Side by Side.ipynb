**Blocking vs Non-Blocking: Side by Side**

The easiest way to understand the difference is to run two tasks that each wait for 2 seconds.

````python
import time

def task1():
    print("Task 1 started")
    time.sleep(2)
    print("Task 1 done")

def task2():
    print("Task 2 started")
    time.sleep(2)
    print("Task 2 done")

start_time = time.time()

task1()
task2()

end_time = time.time()

elapsed = int(end_time - start_time)

print(f"Elapsed: {elapsed} second(s)")
````

**Why?**

time.sleep(2) blocks the current thread.

Task 2 cannot start until Task 1 is completely finished.

🟢 2. Non-blocking version

Now use `asyncio.sleep()` and `asyncio.gather()`

````python
import asyncio

async def task1():
    print("Task 1 started")
    await asyncio.sleep(2)
    print("Task 1 done")

async def task2():
    print("Task 2 started")
    await asyncio.sleep(2)
    print("Task 2 done")

async def main():
    await asyncio.gather(task1(), task2())

start_time = time.time()

await main()

end_time = time.time()

elapsed = int(end_time - start_time)

print(f"Elapsed: {elapsed} second(s)")
````

### **🧠 Side-by-side comparison**

```
        🔴 BLOCKING                    🟢 NON-BLOCKING
        ---------                     --------------

       Task 1                         Task 1
          │                              │
          ▼                              ▼
   time.sleep(2)                 await asyncio.sleep(2)
          │                              │
          │                              └── Give control back
          │                                      │
          │                                      ▼
          │                                  Task 2 runs
          │                                      │
          ▼                                      ▼
     Task 1 done                         Task 2 waiting
          │                                      │
          ▼                                      ▼
       Task 2                             Task 1 resumes
          │                                      │
          ▼                                      ▼
     Task 2 done                         Both complete
     ```
