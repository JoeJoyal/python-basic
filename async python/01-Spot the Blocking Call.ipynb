**input() - Blocking**

````python
def get_name():
    name = input("Enter your name:")
    return name
print("blocking")
````

````python
get_name()
````

**Why?** Python waits for the user to enter something.

**time.sleep() inside a loop - Blocking**

````python
import time

def process():
    for i in range(5):
        print(i)
        time.sleep(1)
print("blocking")        
````

The loop pauses for 1 second during every iteration.

**open().read() - Potentially Blocking I/O**

````python
def read_file():
    with open("large_file.txt", "r") as f:
        data = f.read()
    return data

print("blocking")
````

Reading from a file is synchronous I/O. The thread waits for the read to complete.

**requests.get() - Blocking**

````python
import requests

def get_data():
    response = requests.get("https://example.com")
    return response

print("blocking")
````

The program waits for the HTTP response

**subprocess.run() - Blocking**

````python
import subprocess

def run_command():
    subprocess.run(["python", "--version"])

print("blocking")
````

`subprocess.run()` waits until the external process finishes.

**thread.join() - Blocking**

````python
import threading
import time

def worker():
    time.sleep(3)

t = threading.Thread(target=worker)
t.start()
t.join()

print("blocking")
````

`join()` makes the current thread wait for `t` to finish

**queue.get() - Blocking**

````python
from queue import Queue

q = Queue()

def get_item():
    item = q.get()
    return item

print("blocking")
````

If the queue is empty, `q.get()` waits until an item becomes available.

**socket.recv() - Blocking**

````python
import socket

def receive_data(sock):
    data = sock.recv(1024)
    return data

print("blocking")
````

With a normal blocking socket, `recv()` waits for data.

**asyncio.sleep() - Non blocking**

````python
import asyncio

async def wait():
    await asyncio.sleep(2)

print("non-blocking")
````

Unlike `time.sleep()`, `asyncio.sleep()` allows the event loop to execute other works while waiting.

**asyncio.gather() - Non-blocking from the event loop perspective**

````python
import asyncio

async def task1():
    await asyncio.sleep(2)

async def task2():
    await asyncio.sleep(2)

async def main():
    await asyncio.gather(task1(), task2())

print("non-blocking")
````

The two async tasks can make progress concurrently while they are awaiting.

**`await` an async network call - Non blocking**

````python
import asyncio

async def fetch_data():
    await asyncio.sleep(2)
    return "data"

async def main():
    result = await fetch_data()

print("non-blocking")
````

The important part is that the awaited operation is an **async operation** that yield control to the event loop.

**CPU-havey loop - Blocking**

````python
def calculate():
    total = 0

    for i in range(100_000_000):
        total +=i
    return total

print("blocking")
````

This isn't a traditional I/O blocking call, but it blocks the current thread/event loop because the CPU-heavy computation doesn't yield control.

**`asyncio.to_thread()` - Non-blocking to the event loop**

````python
import asyncio
import time

def blocking_function():
    time.sleep(3)
    return "done"

async def main():
    result = await asyncio.to_thread(blocking_function)

print("non-blocking")
````

Here, `time.sleep(2)` itself is still **blocking,** but it runs in another thread.
Therefore, the asyncio event loop is not blocked

**`time.sleep()` inside an `async def` - still blocking** 

````python
import asyncio
import time

async def bad_task():
    time.sleep(2)

async def main():
    await bad_task()

print("blocking")
````

Even though the function is declared with:

**async def**

the following is still synchronous:

**time.sleep()**

It blocks the event-loop thread

````python
import asyncio
import time

async def good_task():
    await asyncio.sleep(2)

print("non-blocking")
````
