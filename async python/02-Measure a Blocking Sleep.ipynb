````python
import time

start_time = time.time()

time.sleep(1)

end_time = time.time()

elapsed = int(end_time - start_time)

print(f"Elapsed: {elapsed} second(s)")
````

```
start_time
    ↓
time.sleep(1)    ← program blocks for 1 second
    ↓
end_time
    ↓
end_time - start_time
    ↓
int(...)
    ↓
Elapsed: 1 second(s)
```
